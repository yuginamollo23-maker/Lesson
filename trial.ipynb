```python
# Functions is a rule for taking zero or more inputs 
# and returning a corresponding output, we difine it using def

def double(x):
    return x * 2
    

```

```python
print(double(5))
print(double(10))
```

```python
def apply_to_one(f):
    '''calls the function f with 1 as its argument'''
    return f(1)
```

```python
print(apply_to_one(double))
```

```python
#what if f is not double
# a function can take another funcion as input
def add_five(x):
    return x + 5
```

```python
print(apply_to_one(add_five))
```

```python
# a function that creates a function
# functions are like variables you can pass them around

def create_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier

double = create_multiplier(2)
triple = create_multiplier(3)












```

```python
print(double(5))
print(triple(5))
```

```python
# A function that calculates total marks for a whole class list

# Class data - like st pauls exam sheet
students = [
    {'name': 'Asha', 'score': 78},
    
    {'name': 'Brian', 'score': 45},
    {'name': 'Faith', 'score': 92},
    {'name': 'Eugen', 'score': 58},
    {'name': 'Yusuf', 'score': 67}

]
```

```python
def class_average(students):
    total = sum(s['score'] for s in students)
    return total / len(students)

def passed_students(students):
    #returns only names who passed >= 50
    return[s['name'] for s in students if s['score'] >= 50]
```

```python
print(f'Average: {class_average(students)}')
print(f'Passed: {passed_students(students)}')







```

```python
# A function that adds 5 marks to everyone who scored below 50

def add_grace_to_all(students):
    for s in students:
        if s['score']< 50:
            s['score'] += 5
    return students
students = add_grace_to_all(students)
```

```python
print(f'New Average: {class_average(students)}')
print(f'Now Passed: {passed_students(students)}')
print(students)
```

```python
# Sorting the class from topper to lowest

def ranked_class(students):
    # sort from highest to lowest
    return sorted(students, key=lambda s: s['score'], reverse=True)

topper_list = ranked_class(students)
```

```python
print('--- Rank List ---')
for i, s in enumerate(topper_list, start=1):
    print(f'{i}. {s['name']} - {s['score']}')
print(f'\nTopper is: {topper_list[0]['name']} with {topper_list[0]['score']} marks')
```
